```python
def GetPhonemes(hiraganaSent):
    result = []
    for kana in hiraganaSent:
        
        #Remove i if kana is ゃ ゅ ょ
        if kana in ["ゃ", "ゅ", "ょ"]:
            result.pop(-1)
        
        transcription = Hiragana2IPA[kana].copy()
        if len(result) > 0 and result[-1] == "Q":
            result.pop(-1)
            transcription[0] += ":"
            
        result += transcription
    return "".join(result)
        
    
TestingWords = [
    "がっこう", 
    "さくら", 
    "けんやかね",
    "さんばん",
    "さんぽ",   
    "さんねん", 
    "さんかい", 
    "あさ",
    "わかる",
    "かがく",
    "そら",
    "ろく",
    "さんびゃく",
    "さっぱり",
    "さんにん",
    "さんまえ",
    "めいし",
    "にゅうこく",
    "きいた",
    "いっしょにしよう",
]

for word in TestingWords:
    print(f"{word} : {GetPhonemes(word)}")
    
```

### Model A: XLSR-53

```python
from transformers import Wav2Vec2FeatureExtractor, Wav2Vec2PhonemeCTCTokenizer

XLSR_MODEL_ID = "facebook/wav2vec2-xlsr-53-espeak-cv-ft"

print(f"Loading {XLSR_MODEL_ID} ...")
xlsr_extractor = Wav2Vec2FeatureExtractor.from_pretrained(XLSR_MODEL_ID)
xlsr_tokenizer = Wav2Vec2PhonemeCTCTokenizer.from_pretrained(XLSR_MODEL_ID)
xlsr_model = Wav2Vec2ForCTC.from_pretrained(XLSR_MODEL_ID)
xlsr_model.eval()
print("XLSR-53 Phoneme model loaded.")
```

```python
import librosa


def RecognizeXLSR(path, extractor=xlsr_extractor, model=xlsr_model, tokenizer=xlsr_tokenizer):
    audio, _ = librosa.load(path, sr=16000, mono=True)

    return RecognizeXLSRAudio(audio, extractor, model, tokenizer)

def RecognizeXLSRAudio(audio, extractor=xlsr_extractor, model=xlsr_model, tokenizer=xlsr_tokenizer):
    inputs = extractor(
        audio,
        sampling_rate=16000,
        return_tensors="pt",
        padding=True
    )

    with torch.no_grad():
        logits = model(**inputs).logits

    predicted_ids = torch.argmax(logits, dim=-1)
    raw = tokenizer.batch_decode(predicted_ids)[0]

    return raw
```

### Model B: Hubert Base Phoneme CTC 

```python
from transformers import HubertForCTC, Wav2Vec2Processor
import numpy as np

HUBERT_BASE_MODEL_ID = "prj-beatrice/japanese-hubert-base-phoneme-ctc"
hubert_ctc_model = HubertForCTC.from_pretrained(HUBERT_BASE_MODEL_ID)
hubert_ctc_processor = Wav2Vec2Processor.from_pretrained(HUBERT_BASE_MODEL_ID)

print("Hubert Base Phoneme CTC Phoneme model loaded.")

def RecognizeHubert(path, processor=hubert_ctc_processor, model=hubert_ctc_model):
    audio, sr = librosa.load(path, sr=16000)
    audio = np.concatenate([np.zeros(sr), audio, np.zeros(sr // 2)])

    return RecognizeHubertAudio(audio, processor, model)

def RecognizeHubertAudio(audio, processor=hubert_ctc_processor, model=hubert_ctc_model):
    inputs = processor(audio, sampling_rate=16000, return_tensors="pt")
    with torch.no_grad():
        outputs = model(**inputs)
        predicted_ids = outputs.logits.argmax(-1)
        phonemes = processor.decode(predicted_ids[0], spaces_between_special_tokens=True)

    return phonemes
    
```

```python
from allosaurus.app import read_recognizer as allo_load
import tempfile

allo_model = allo_load()
print("Allosaurus model loaded.")

def RecognizeAllosaurus(path, lang="jpn"):
    audio, sr = librosa.load(path, sr=16000)
    audio = np.concatenate([np.zeros(sr), audio, np.zeros(sr // 2)])

    return RecognizeAllosaurusAudio(audio, lang)


def RecognizeAllosaurusAudio(audio, lang="jpn"):
    with tempfile.NamedTemporaryFile(suffix=".wav") as f:
        sf.write(f.name, audio, 16000)

        raw = allo_model.recognize(f.name, lang)

    phones = [p for p in raw.strip().split() if p]
    return " ".join(phones)
```

### Model D: Facebook MMS-1B-All

```python
from transformers import Wav2Vec2ForCTC, AutoProcessor

MMS_MODEL_ID = "facebook/mms-1b-all"

mms_processor = AutoProcessor.from_pretrained(MMS_MODEL_ID)
mms_model = Wav2Vec2ForCTC.from_pretrained(MMS_MODEL_ID)


def RecognizeMMS(path, processor=mms_processor, model=mms_model):
    
    audio, sr = librosa.load(path, sr=16000)
    audio = np.concatenate([np.zeros(sr), audio, np.zeros(sr // 2)])
    return RecognizeMMSAudio(audio, processor, model)
    

def RecognizeMMSAudio(audio, processor=mms_processor, model=mms_model):
    inputs = processor(audio, sampling_rate=16_000, return_tensors="pt")

    with torch.no_grad():
        outputs = model(**inputs).logits

    ids = torch.argmax(outputs, dim=-1)[0]
    transcription = processor.decode(ids)
    return " ".join(transcription)
    
```

```python
import jiwer

def PER(refs, preds):
    per = jiwer.wer(refs, preds)
    return per

def LER(refs, preds):
    totalRef = 0
    totalError = 0

    for ref, pred in zip(refs, preds):
        rLen = len(ref.split())
        pLen = len(pred.split())

        totalRef += rLen
        totalError += abs(pLen - rLen)

    return totalError / totalRef


def MLR(refs, preds):
    totalRef = 0
    totalPred = 0

    for ref, pred in zip(refs, preds):
        totalRef += len(ref.split())
        totalPred += len(pred.split())

    return totalPred / totalRef


for model in models:
    print(f"\n=== {model} ===")
    
    print(f"{model} PER (US)  : {PER(test_df["IPA"].to_list()[:670], transcriptions[model][:670])}")
    print(f"{model} PER (JP)  : {PER(test_df["IPA"].to_list()[670:], transcriptions[model][670:])}")
    print(f"{model} PER (ALL) : {PER(test_df["IPA"].to_list(), transcriptions[model])}")
    
```

# Feature-Level Evaluation


```python
import pandas as pd


def AlignPhonemes(ref, pred):
    n, m = len(ref), len(pred)
    dp = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(n + 1):
        dp[i][0] = i
    for j in range(m + 1):
        dp[0][j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            sub = dp[i-1][j-1] + (ref[i-1] != pred[j-1])
            dp[i][j] = min(sub, dp[i-1][j] + 1, dp[i][j-1] + 1)

    pairs = []
    i, j = n, m
    while i > 0 or j > 0:
        if i > 0 and j > 0 and dp[i][j] == dp[i-1][j-1] + (ref[i-1] != pred[j-1]):
            pairs.append((ref[i-1], pred[j-1]))
            i, j = i - 1, j - 1
        elif i > 0 and dp[i][j] == dp[i-1][j] + 1:
            pairs.append((ref[i-1], None)) #Deletion
            i -= 1
        else:
            pairs.append((None, pred[j-1])) #Insertion
            j -= 1
    return pairs[::-1]


R_FEATURE = {
    "name": "R",
    "targets": {("jp", "ɾ"): "ɾ", ("us", "ɾ"): "ɾ", ("us", "r"): "ɹ"},
    "correct": {"ɾ": "ɾ", "ɹ": "ɹ"},
    "categories": ["ɾ", "ɹ", "other"],
    "symbols": {"ɾ": "ɾ", "ɾʲ": "ɾ", "ɹ": "ɹ", "ɻ": "ɹ"},
    "modelSymbols": {
        "hubert":     {"r": "ɾ", "ry": "ɾ"}, #Japanese-only inventory
        "fine-tuned": {"r": "ɹ"},            #Learned from TIMIT r
        "xlsr": {"r": "ɾ"},
    },
}


def FeatureConfusion(feature, model):
    #Optional hooks: "preprocess" rewrites the reference and model tokens before alignment,
    #"classify" replaces the symbols/modelSymbols lookup with a function
    preprocess = feature.get("preprocess", lambda tokens: tokens)
    if "classify" in feature:
        classify = feature["classify"]
    else:
        symbols = {**feature["symbols"], **feature["modelSymbols"].get(model, {})}
        classify = lambda symbol: symbols.get(symbol, "other")

    rows = []
    for ref, pred, lang in zip(test_df["IPA"], transcriptions[model], test_df["lang"]):
        for refPhone, predPhone in AlignPhonemes(preprocess(ref.split()), preprocess(pred.split())):
            target = feature["targets"].get((lang, refPhone))
            if target is None:
                continue
            rows.append({"model": model, "lang": lang, "target": target, "reference": refPhone, "produced": classify(predPhone), "symbol": predPhone})
    return rows


rDf = pd.DataFrame([row for model in models for row in FeatureConfusion(R_FEATURE, model)])
rDf["correct"] = rDf["produced"] == rDf["target"].map(R_FEATURE["correct"])

print(rDf[rDf["model"] == models[0]].groupby(["target", "lang"]).size())
```

#### Confusion matrices
Rows are normalised: each cell is the % of that target's occurrences produced as that category. Outlined cells are the correct answer.
"other" covers every other phoneme and deletions.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches


def PlotFeatureConfusion(featureDf, feature):
    targets = list(dict.fromkeys(feature["targets"].values()))
    categories = feature["categories"]

    fig, axes = plt.subplots(1, len(models), figsize=(3.2 * len(models), 3.4), sharey=True)
    for ax, model in zip(axes, models):
        sub = featureDf[featureDf["model"] == model]
        matrix = pd.crosstab(sub["target"], sub["produced"], normalize="index")
        matrix = matrix.reindex(index=targets, columns=categories, fill_value=0)

        ax.imshow(matrix.values, cmap="Blues", vmin=0, vmax=1)
        for i, target in enumerate(targets):
            for j, category in enumerate(categories):
                value = matrix.iat[i, j]
                ax.text(j, i, f"{value:.0%}", ha="center", va="center", fontsize=10,
                        color="white" if value > 0.6 else "#222222")
                if category == feature["correct"][target]:
                    ax.add_patch(mpatches.Rectangle((j - 0.5, i - 0.5), 1, 1, fill=False, edgecolor="#222222", linewidth=1.5))

        ax.set_xticks(range(len(categories)))
        ax.set_xticklabels(categories)
        ax.set_xlabel("Produced")
        ax.set_title(model, fontsize=11)

    counts = featureDf[featureDf["model"] == models[0]]["target"].value_counts()
    axes[0].set_yticks(range(len(targets)))
    axes[0].set_yticklabels([f"{t} (n={counts.get(t, 0)})" for t in targets])
    fig.suptitle(f"{feature['name']}: target → produced", fontsize=13)
    plt.tight_layout()
    plt.show()


PlotFeatureConfusion(rDf, R_FEATURE)
```

#### Detection accuracy per model
% of each target produced as the correct category, i.e. the outlined diagonal of the matrices above.
A model is only useful for this feature if it gets *both* ɾ and ɹ right.

```python
def PlotFeatureAccuracy(featureDf, feature, colors):
    targets = list(dict.fromkeys(feature["targets"].values()))
    accuracy = featureDf.pivot_table(index="model", columns="target", values="correct", aggfunc="mean")
    accuracy = accuracy.reindex(index=models, columns=targets)
    #Occurrences come from the references, so they are the same for every model
    counts = featureDf[featureDf["model"] == models[0]]["target"].value_counts()

    x = np.arange(len(models))
    width = 0.8 / len(targets)

    _, ax = plt.subplots(figsize=(9, 5))
    for k, target in enumerate(targets):
        offset = (k - (len(targets) - 1) / 2) * width
        ax.bar(x + offset, accuracy[target], width, label=f"{target} (n={counts.get(target, 0)})", color=colors[target], edgecolor="white", linewidth=1)

    for container in ax.containers:
        ax.bar_label(container, fmt="%.2f", padding=3, fontsize=8)

    ax.set_xticks(x)
    ax.set_xticklabels(models)
    ax.set_ylim(0, 1.1)
    ax.set_ylabel("Produced correctly")
    ax.set_xlabel("Model")
    ax.set_title(f"{feature['name']} detection accuracy X Model")
    ax.legend()

    plt.tight_layout()
    plt.show()
    return accuracy


LANG_NAMES = {"jp": "JP", "us": "EN"}


def PlotFeatureAccuracyByLang(featureDf, feature, colors):
    targets = list(dict.fromkeys(feature["targets"].values()))
    counts = featureDf[featureDf["model"] == models[0]].groupby(["target", "lang"]).size()
    #Every target x language slot is drawn; slots with no references are left empty (n/a)
    groups = [(t, lang) for t in targets for lang in LANG_NAMES]
    accuracy = featureDf.pivot_table(index="model", columns=["target", "lang"], values="correct", aggfunc="mean")
    accuracy = accuracy.reindex(index=models, columns=groups)

    x = np.arange(len(models))
    width = 0.8 / len(groups)

    _, ax = plt.subplots(figsize=(9, 5))
    for k, (target, lang) in enumerate(groups):
        offset = (k - (len(groups) - 1) / 2) * width
        n = counts.get((target, lang), 0)
        values = accuracy[(target, lang)].fillna(0)
        bars = ax.bar(x + offset, values, width, label=f"{target} · {LANG_NAMES[lang]} (n={n})",
               color=colors[target], hatch="//" if lang == "us" else None, edgecolor="white", linewidth=1)
        ax.bar_label(bars, labels=[f"{v:.2f}" if n else "n/a" for v in values], padding=3, fontsize=8)

    ax.set_xticks(x)
    ax.set_xticklabels(models)
    ax.set_ylim(0, 1.25)
    ax.set_ylabel("Produced correctly")
    ax.set_xlabel("Model")
    ax.set_title(f"{feature['name']} detection accuracy X Model, by sentence language")
    ax.legend(loc="upper center", ncol=len(groups), fontsize=9) #One row above the bars

    plt.tight_layout()
    plt.show()
    return accuracy


rAccuracy = PlotFeatureAccuracy(rDf, R_FEATURE, colors={"ɾ": "#ef3948", "ɹ": "#5faeeb"})
PlotFeatureAccuracyByLang(rDf, R_FEATURE, colors={"ɾ": "#ef3948", "ɹ": "#5faeeb"})
rAccuracy.round(3)
```

### Feature 2: Geminate vs singleton consonants

```python
GEMINATES = {"kː", "tː", "pː", "sː", "ɕː", "tsː", "tɕː", "dː", "ɡː", "dʑː", "hː"}
SINGLETONS = {g[:-1] for g in GEMINATES} | {"g"} 

GEMINABLE = SINGLETONS | {"t̪", "s̪", "d̪", "ʃ", "sh", "tʃ", "ch", "dʒ", "b", "z", "f"}
VOWEL_CHARS = set("aeiouɑɯəɛɔɪʊʌæɐɚɝøœɵɨʉAIUEO")


def MergeGeminates(tokens):
    merged = []
    for token in tokens:
        prev = merged[-1] if merged else None
        if token in GEMINABLE and prev in ("cl", token): #Hubert "cl t" / XLSR "t t"
            merged[-1] = token + "ː"
        elif prev in GEMINABLE and token == prev + "ː":  #"k kː"
            merged[-1] = token
        else:
            merged.append(token)
    return merged


def ClassifyLength(symbol):
    if symbol is None or not symbol[0].isalpha() or symbol[0] in VOWEL_CHARS:
        return "other"
    if symbol == "cl" or symbol.endswith("ː"):
        return "geminate"
    return "singleton"


GEMINATE_FEATURE = {
    "name": "Gemination",
    "targets": {("jp", g): "geminate" for g in GEMINATES}
             | {(lang, s): "singleton" for lang in ["jp", "us"] for s in SINGLETONS},
    "correct": {"geminate": "geminate", "singleton": "singleton"},
    "categories": ["geminate", "singleton", "other"],
    "preprocess": MergeGeminates,
    "classify": ClassifyLength,
}

gDf = pd.DataFrame([row for model in models for row in FeatureConfusion(GEMINATE_FEATURE, model)])
gDf["correct"] = gDf["produced"] == gDf["target"].map(GEMINATE_FEATURE["correct"])

PlotFeatureConfusion(gDf, GEMINATE_FEATURE)
gAccuracy = PlotFeatureAccuracy(gDf, GEMINATE_FEATURE, colors={"geminate": "#ef3948", "singleton": "#5faeeb"})
PlotFeatureAccuracyByLang(gDf, GEMINATE_FEATURE, colors={"geminate": "#ef3948", "singleton": "#5faeeb"})
gAccuracy.round(3)
```

### Feature 3: Devoiced vs voiced vowels

```python
DEVOICED = {"i̥", "ɯ̥", "I", "U"} #Hubert marks devoiced vowels with capitals


def ClassifyVoicing(symbol):
    if symbol in DEVOICED:
        return "devoiced"
    if symbol is not None and (symbol[0] in VOWEL_CHARS or symbol == "y"): #XLSR y is the vowel [y]
        return "voiced"
    return "dropped" #Deleted, or the aligner put a neighbouring consonant in the vowel's slot


DEVOICING_FEATURE = {
    "name": "Devoicing",
    "targets": {("jp", "i̥"): "devoiced", ("jp", "ɯ̥"): "devoiced", ("jp", "i"): "voiced", ("jp", "ɯ"): "voiced",
                ("us", "i"): "voiced", ("us", "u"): "voiced"}, #English high vowels (never devoiced in TIMIT)
    "correct": {"devoiced": "devoiced", "voiced": "voiced"},
    "categories": ["devoiced", "voiced", "dropped"],
    "classify": ClassifyVoicing,
}

dDf = pd.DataFrame([row for model in models for row in FeatureConfusion(DEVOICING_FEATURE, model)])
dDf["correct"] = dDf["produced"] == dDf["target"].map(DEVOICING_FEATURE["correct"])

PlotFeatureConfusion(dDf, DEVOICING_FEATURE)
dAccuracy = PlotFeatureAccuracy(dDf, DEVOICING_FEATURE, colors={"devoiced": "#ef3948", "voiced": "#5faeeb"})
PlotFeatureAccuracyByLang(dDf, DEVOICING_FEATURE, colors={"devoiced": "#ef3948", "voiced": "#5faeeb"})
dAccuracy.round(3)
```

### Feature 4: Long vs short vowels

```python
JP_VOWELS = ["a", "i", "ɯ", "e", "o"]
EN_VOWELS = ["i", "ɪ", "u", "ʊ", "ɛ", "æ", "ɑ", "ɔ", "ʌ", "ə"] #TIMIT monophthongs, no length marking so all short


def IsVowel(symbol):
    return symbol is not None and (symbol[0] in VOWEL_CHARS or symbol == "y") #XLSR y is the vowel [y]


def MergeLongVowels(tokens):
    merged = []
    for token in tokens:
        prev = merged[-1] if merged else None
        if IsVowel(token) and prev is not None and token.rstrip("ː") == prev.rstrip("ː"): #JVS/Hubert "o o", XLSR "oː o"
            merged[-1] = token.rstrip("ː") + "ː"
        else:
            merged.append(token)
    return merged


def ClassifyVowelLength(symbol):
    if not IsVowel(symbol):
        return "other"
    return "long" if symbol.endswith("ː") else "short"


LONG_VOWEL_FEATURE = {
    "name": "Vowel length",
    "targets": {("jp", v + "ː"): "long" for v in JP_VOWELS} | {("jp", v): "short" for v in JP_VOWELS}
             | {("us", v): "short" for v in EN_VOWELS},
    "correct": {"long": "long", "short": "short"},
    "categories": ["long", "short", "other"],
    "preprocess": MergeLongVowels,
    "classify": ClassifyVowelLength,
}

lDf = pd.DataFrame([row for model in models for row in FeatureConfusion(LONG_VOWEL_FEATURE, model)])
lDf["correct"] = lDf["produced"] == lDf["target"].map(LONG_VOWEL_FEATURE["correct"])

PlotFeatureConfusion(lDf, LONG_VOWEL_FEATURE)
lAccuracy = PlotFeatureAccuracy(lDf, LONG_VOWEL_FEATURE, colors={"long": "#ef3948", "short": "#5faeeb"})
PlotFeatureAccuracyByLang(lDf, LONG_VOWEL_FEATURE, colors={"long": "#ef3948", "short": "#5faeeb"})
lAccuracy.round(3)
```
