#### Hiragana to IPA Dictionary

```python
Hiragana2IPA = {
    "あ": ["ɑ̟"],
    "い": ["i"],
    "う": ["ɯ̟"],
    "え": ["e̞"],
    "お": ["o̞"],
    
    # KA KI KU KE KO
    "か": ["k","ɑ̟"],
    "き": ["kʲ","i"],
    "く": ["k","ɯ̟"],
    "け": ["k","e̞"],
    "こ": ["k","o̞"],

    # GA GI GU GE GO
    "が": ["ɡ","ɑ̟"],
    "ぎ": ["ɡʲ","i"],
    "ぐ": ["ɡ","ɯ̟"],
    "げ": ["ɡ","e̞"],
    "ご": ["ɡ","o̞"],
    
    # SA SHI SU SE SO
    "さ": ["s","ɑ̟"],
    "し": ["ɕ","i"],
    "す": ["s","ɯ̟"],
    "せ": ["s","e̞"],
    "そ": ["s","o̞"],

    # ZA JI ZU ZE ZO
    "ざ": ["z","ɑ̟"],
    "じ": ["dʑ","i"],
    "ず": ["z","ɯ̟"],
    "ぜ": ["z","e̞"],
    "ぞ": ["z","o̞"],

    # TA CHI TSU TE TO
    "た": ["t","ɑ̟"],
    "ち": ["tɕ","i"],
    "つ": ["ts","ɯ̟"],
    "て": ["t","e̞"],
    "と": ["t","o̞"],

    # DA DI DU DE DO
    "だ": ["d","ɑ̟"],
    "ぢ": ["dʑ","i"],
    "づ": ["dʑ","ɯ̟"], 
    "で": ["d","e̞"],
    "ど": ["d","o̞"],

    # NA NI NU NE NO
    "な": ["n","ɑ̟"],
    "に": ["ɲ","i"],
    "ぬ": ["n","ɯ̟"],
    "ね": ["n","e̞"],
    "の": ["n","o̞"],

    # HA HI FU HE HO
    "は": ["h","ɑ̟"],
    "ひ": ["ç","i"],
    "ふ": ["ɸ","ɯ̟"],
    "へ": ["h","e̞"],
    "ほ": ["h","o̞"],

    # BA BI BU BE BO
    "ば": ["b","ɑ̟"],
    "び": ["b","i"],
    "ぶ": ["b","ɯ̟"],
    "べ": ["b","e̞"],
    "ぼ": ["b","o̞"],

    # PA PI PU PE PO
    "ぱ": ["p","ɑ̟"],
    "ぴ": ["p","i"],
    "ぷ": ["p","ɯ̟"],
    "ぺ": ["p","e̞"],
    "ぽ": ["p","o̞"],

    # MA MI MU ME MO
    "ま": ["m","ɑ̟"],
    "み": ["m","i"],
    "む": ["m","ɯ̟"],
    "め": ["m","e̞"],
    "も": ["m","o̞"],

    # YA YU YO
    "や": ["j","ɑ̟"],
    "ゆ": ["j","ɯ̟"],
    "よ": ["j","o̞"],

    # RA RI RU RE RO
    "ら": ["ɾ","ɑ̟"],
    "り": ["ɾʲ","i"],
    "る": ["ɾ","ɯ̟"],
    "れ": ["ɾ","e̞"],
    "ろ": ["ɾ","o̞"],

    # WA
    "わ": ["w","ɑ̟"],

    # Moraic nasal
    "ん": ["N"],

    # Small kana
    "ぁ": ["ɑ̟"],
    "ぃ": ["i"],
    "ぅ": ["ɯ̟"],
    "ぇ": ["e̞"],
    "ぉ": ["o̞"],

    "ゃ": ["j","ɑ̟"],
    "ゅ": ["j","ɯ̟"],
    "ょ": ["j","o̞"],

    # Sokuon
    "っ": ["Q"],
}
```

```python
def GetPhonemes(hiraganaSent):
    result = []
    for kana in hiraganaSent:
        
        #Remove i if kana is ゃ ゅ ょ
        if kana in ["ゃ", "ゅ", "ょ"]:
            result.pop(-1)
        
        transcription = Hiragana2IPA[kana].copy()
        if len(result) > 0 and result[-1] == "Q":
            result.pop(-1)
            transcription[0] += ":"
            
        result += transcription
    return "".join(result)
        
    
TestingWords = [
    "がっこう", 
    "さくら", 
    "けんやかね",
    "さんばん",
    "さんぽ",   
    "さんねん", 
    "さんかい", 
    "あさ",
    "わかる",
    "かがく",
    "そら",
    "ろく",
    "さんびゃく",
    "さっぱり",
    "さんにん",
    "さんまえ",
    "めいし",
    "にゅうこく",
    "きいた",
    "いっしょにしよう",
]

for word in TestingWords:
    print(f"{word} : {GetPhonemes(word)}")
    
```

### Model A: XLSR-53

```python
from transformers import Wav2Vec2FeatureExtractor, Wav2Vec2PhonemeCTCTokenizer

XLSR_MODEL_ID = "facebook/wav2vec2-xlsr-53-espeak-cv-ft"

print(f"Loading {XLSR_MODEL_ID} ...")
xlsr_extractor = Wav2Vec2FeatureExtractor.from_pretrained(XLSR_MODEL_ID)
xlsr_tokenizer = Wav2Vec2PhonemeCTCTokenizer.from_pretrained(XLSR_MODEL_ID)
xlsr_model = Wav2Vec2ForCTC.from_pretrained(XLSR_MODEL_ID)
xlsr_model.eval()
print("XLSR-53 Phoneme model loaded.")
```

```python
import librosa


def RecognizeXLSR(path, extractor=xlsr_extractor, model=xlsr_model, tokenizer=xlsr_tokenizer):
    audio, _ = librosa.load(path, sr=16000, mono=True)

    return RecognizeXLSRAudio(audio, extractor, model, tokenizer)

def RecognizeXLSRAudio(audio, extractor=xlsr_extractor, model=xlsr_model, tokenizer=xlsr_tokenizer):
    inputs = extractor(
        audio,
        sampling_rate=16000,
        return_tensors="pt",
        padding=True
    )

    with torch.no_grad():
        logits = model(**inputs).logits

    predicted_ids = torch.argmax(logits, dim=-1)
    raw = tokenizer.batch_decode(predicted_ids)[0]

    return raw
```

### Model B: Hubert Base Phoneme CTC 

```python
from transformers import HubertForCTC, Wav2Vec2Processor
import numpy as np

HUBERT_BASE_MODEL_ID = "prj-beatrice/japanese-hubert-base-phoneme-ctc"
hubert_ctc_model = HubertForCTC.from_pretrained(HUBERT_BASE_MODEL_ID)
hubert_ctc_processor = Wav2Vec2Processor.from_pretrained(HUBERT_BASE_MODEL_ID)

print("Hubert Base Phoneme CTC Phoneme model loaded.")

def RecognizeHubert(path, processor=hubert_ctc_processor, model=hubert_ctc_model):
    audio, sr = librosa.load(path, sr=16000)
    audio = np.concatenate([np.zeros(sr), audio, np.zeros(sr // 2)])

    return RecognizeHubertAudio(audio, processor, model)

def RecognizeHubertAudio(audio, processor=hubert_ctc_processor, model=hubert_ctc_model):
    inputs = processor(audio, sampling_rate=16000, return_tensors="pt")
    with torch.no_grad():
        outputs = model(**inputs)
        predicted_ids = outputs.logits.argmax(-1)
        phonemes = processor.decode(predicted_ids[0], spaces_between_special_tokens=True)

    return phonemes
    
```

```python
from allosaurus.app import read_recognizer as allo_load
import tempfile

allo_model = allo_load()
print("Allosaurus model loaded.")

def RecognizeAllosaurus(path, lang="jpn"):
    audio, sr = librosa.load(path, sr=16000)
    audio = np.concatenate([np.zeros(sr), audio, np.zeros(sr // 2)])

    return RecognizeAllosaurusAudio(audio, lang)


def RecognizeAllosaurusAudio(audio, lang="jpn"):
    with tempfile.NamedTemporaryFile(suffix=".wav") as f:
        sf.write(f.name, audio, 16000)

        raw = allo_model.recognize(f.name, lang)

    phones = [p for p in raw.strip().split() if p]
    return " ".join(phones)
```

### Model E: Facebook MMS-1B-All

```python
from transformers import Wav2Vec2ForCTC, AutoProcessor

MMS_MODEL_ID = "facebook/mms-1b-all"

mms_processor = AutoProcessor.from_pretrained(MMS_MODEL_ID)
mms_model = Wav2Vec2ForCTC.from_pretrained(MMS_MODEL_ID)


def RecognizeMMS(path, processor=mms_processor, model=mms_model):
    
    audio, sr = librosa.load(path, sr=16000)
    audio = np.concatenate([np.zeros(sr), audio, np.zeros(sr // 2)])
    return RecognizeMMSAudio(audio, processor, model)
    

def RecognizeMMSAudio(audio, processor=mms_processor, model=mms_model):
    inputs = processor(audio, sampling_rate=16_000, return_tensors="pt")

    with torch.no_grad():
        outputs = model(**inputs).logits

    ids = torch.argmax(outputs, dim=-1)[0]
    transcription = processor.decode(ids)
    return " ".join(transcription)
    
```

```python
import io 

@torch.no_grad()
def Transcribe(model, audio, sr=16000) -> str:
    
    if isinstance(audio, dict):
        audio, sr = sf.read(io.BytesIO(audio["bytes"]), dtype="float32")
    
    audio = np.asarray(audio, dtype=np.float32)
    if audio.ndim > 1: #If stero put to mono
        audio = audio.mean(axis=1)
        
    if model == "xlsr":
        return RecognizeXLSRAudio(audio)
    elif model == "hubert":
        return RecognizeHubertAudio(audio)
    elif model == "allosaurus":
        return RecognizeAllosaurusAudio(audio)
    elif model == "mms":
        return RecognizeMMSAudio(audio)
    
@torch.no_grad()
def TranscribeColumn(model, testingCol):
    return testingCol.apply(lambda audio: Transcribe(model, audio))
    
```

```python
import jiwer

def PER(refs, preds):
    per = jiwer.wer(refs, preds)
    return per

def LER(refs, preds):
    totalRef = 0
    totalError = 0

    for ref, pred in zip(refs, preds):
        rLen = len(ref.split())
        pLen = len(pred.split())

        totalRef += rLen
        totalError += abs(pLen - rLen)

    return totalError / totalRef


def MLR(refs, preds):
    totalRef = 0
    totalPred = 0

    for ref, pred in zip(refs, preds):
        totalRef += len(ref.split())
        totalPred += len(pred.split())

    return totalPred / totalRef


for model in models:
    print(f"\n=== {model} ===")
    
    print(f"{model} PER (US)  : {PER(test_df["IPA"].to_list()[:670], transcriptions[model][:670])}")
    print(f"{model} PER (JP)  : {PER(test_df["IPA"].to_list()[670:], transcriptions[model][670:])}")
    print(f"{model} PER (ALL) : {PER(test_df["IPA"].to_list(), transcriptions[model])}")
    
```

```python
import matplotlib.pyplot as plt
import numpy as np

def PlotBarChart(overall, us, jp, title, metric):

    x = np.arange(len(models))
    width = 0.25

    _, ax = plt.subplots(figsize=(8, 5))

    ax.bar(x - width, overall, width, label="Overall", color="#ffb300")
    ax.bar(x, us, width, label="US", color="#5faeeb")
    ax.bar(x + width, jp, width, label="Japanese", color="#ef3948")

    ax.set_xticks(x)
    ax.set_xticklabels(models)
    ax.set_ylabel(metric)
    ax.set_xlabel("Model")
    ax.set_title(title)
    ax.legend()

    for container in ax.containers:
        ax.bar_label(container, fmt="%.3f", padding=3)

    plt.tight_layout()
    plt.show()

def PlotBarChartCentered(overall, us, jp, title, metric):
    x = np.arange(len(models))
    width = 0.25

    fig, ax = plt.subplots(figsize=(8, 5))

    baseline = 1.0

    overall_h = np.array(overall) - baseline
    us_h = np.array(us) - baseline
    jp_h = np.array(jp) - baseline

    ax.bar(x - width, overall_h, width, bottom=baseline, label="Overall", color="#ffb300")
    ax.bar(x, us_h, width, bottom=baseline, label="US", color="#5faeeb")
    ax.bar(x + width, jp_h, width, bottom=baseline, label="Japanese", color="#ef3948")

    all_values = np.concatenate([overall, us, jp])
    max_dev = np.max(np.abs(all_values - baseline))
    margin = max_dev * 0.15

    ax.set_ylim(baseline - max_dev - margin, baseline + max_dev + margin)

    ax.axhline(baseline, color="black", linewidth=1.5)

    ax.set_xticks(x)
    ax.set_xticklabels(models)
    ax.set_ylabel(metric)
    ax.set_xlabel("Model")
    ax.set_title(title)
    ax.legend()

    for container in ax.containers:
        labels = [f"{b.get_y() + b.get_height():.3f}" for b in container]
        ax.bar_label(container, labels=labels, padding=3)

    plt.tight_layout()
    plt.show()
    
metrics = [
    {
        "function" : PER,
        "metric" : "Phoneme Error Rate (PER)",
        "title" : "PER X Model",
        "plot" : PlotBarChart,
    },
    {
        "function" : LER,
        "metric" : "Length Error Rate (LER)",
        "title" : "LER X Model",
        "plot" : PlotBarChart,
    },
    {
        "function" : MLR,
        "metric" : "Mean Length Ratio (MLR)",
        "title" : "MLR X Model",
        "plot" : PlotBarChartCentered
    }
]

for metric in metrics:
    overallPER = [metric["function"](test_df["IPA"].to_list(), transcriptions[model]) for model in models]
    usPER = [metric["function"](test_df["IPA"].to_list()[:670], transcriptions[model][:670]) for model in models]
    jpPER = [metric["function"](test_df["IPA"].to_list()[670:], transcriptions[model][670:]) for model in models]
        
    metric["plot"](overallPER, usPER, jpPER, metric["title"], metric["metric"])

```
